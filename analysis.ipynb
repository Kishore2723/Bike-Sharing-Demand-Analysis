### Cleaning the data

```python
# Missing values checking

df.isnull().sum()
```

```python
# Duplicate Checking
df.duplicated().sum()
```

```python
# Data format correction Data type correction
df['Date'] = pd.to_datetime(df['Date'], format='%d-%m-%Y')
```

```python
# Category
df['Seasons'].unique()
df['Holiday'].unique()
df['Functioning Day'].unique()

# Hours should be less than 24
df[(df['Hour'] < 0) | (df['Hour'] > 23)]

# Humidity Should be without negative values
df[(df['Humidity(%)'] < 0) | (df['Humidity(%)'] > 100)]
```

```python
df.info()
```

### Create Time Category 

```python
# Category 
def time_category(hour):
    if 5 <= hour <= 11:
        return 'Morning'
    elif 12 <= hour <= 16:
        return 'Afternoon'
    elif 17 <= hour <= 21:
        return 'Evening'
    else:
        return 'Night'

df['Time_Category'] = df['Hour'].apply(time_category)

df.head(10)
```

```python
df['Time_Category'].value_counts()
```

###  Create Weather Category 

```python
def weather_category(rainfall):
    if rainfall == 0:
        return "No Rain"
    elif rainfall <= 5:
        return "Low Rain"
    else:
        return "Heavy Rain"

df['Weather_Category'] = df['Rainfall(mm)'].apply(weather_category)

df.head(5)
```

### Basic Analysis

```python
# Total Bike rentals

print("The total count of Rented bike: ",df['Rented Bike Count'].sum())
```

```python
# Average Hourly Rentals

hourly_avg = df.groupby('Hour')['Rented Bike Count'].mean()
print(hourly_avg)

hourly_avg.plot(kind='line')
plt.xlabel('Hour')
plt.ylabel('Average Bike Rentals')
plt.title('Average Bike Rentals by Hour')
plt.show()
```

```python
print("Hour: ",df.groupby('Hour')['Rented Bike Count'].mean().idxmax())
print("Maximum number of bikes in a hour: ",df.groupby('Hour')['Rented Bike Count'].mean().max())
```

```python
# Highest demand season
season_avg = df.groupby('Seasons')['Rented Bike Count'].mean()
print(season_avg)
season_avg.plot(kind='bar')
plt.xlabel('Season')
plt.ylabel('Average Bike Rentals')
plt.title('Average Bike Rentals by Season')
plt.show()
```

```python
# Holiday Vs Non-Holiday
holiday_avg = df.groupby('Holiday')['Rented Bike Count'].mean()
print(holiday_avg)
holiday_avg.plot(kind='bar')
plt.xlabel('Holiday Status')
plt.ylabel('Average Bike Rentals')
plt.title('Holiday vs Non-Holiday Bike Rentals')
plt.show()
```

```python
# Rainy Vs Non-Rainy 

def rainy_category(rainfall):
    if rainfall > 0:
        return "Rainy"
    else:
        return "Non Rainy"

df['Rainy_Category'] = df['Rainfall(mm)'].apply(rainy_category)


rainy_avg = df.groupby('Rainy_Category')['Rented Bike Count'].mean()
print(rainy_avg)


plt.bar(rainy_avg.index, rainy_avg.values)
plt.xlabel('Rainy Category')
plt.ylabel('Average Rented Bikes')
plt.title('Rainy vs Non-Rainy Bike Rentals')
plt.show()
```

```python
# Cleaned Dataset

df.to_csv("cleaned_bike_data.csv", index=False)
```

## Web Scrapping

### Importing libraries

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

### Web page Loading

```python
# Get the Web Page

url = "https://english.visitseoul.net/attractions"
response = requests.get(url)
print(response.status_code)
```

### Create BeautifulSoup object

```python
soup = BeautifulSoup(response.text, 'html.parser')
```

```python
# Imspecting the Web Page

print(soup.title)
```

```python
# For Attraction extraction

attractions = []

for link in soup.find_all('a'):
    name = link.get_text(strip=True)
    
    if name:
        attractions.append(name)

for attraction in attractions:
    print(attraction)
```

```python
categories = [
    'Landmark',
    'Shopping',
    'Historical',
    'Historical',
    'Historical',
    'Landmark',
    'Historical',
    'Landmark'
]

seoul_attractions = [
    'Namsan Seoul Tower',
    'Starfield Library',
    'Gyeongbokgung Palace',
    'Hanyangdoseong (Seoul City Wall)',
    'Changdeokgung Palace',
    'Banpodaegyo Bridge Moonlight Rainbow Fountain',
    'Bukchon Hanok Village',
    'Lotte World Tower'
]

df_attractions = pd.DataFrame({
    'Category or Area': categories,
    'Attraction': seoul_attractions
})

print(df_attractions)
```

```python
df_attractions.to_csv('scraped_data.csv', index=False)
```
